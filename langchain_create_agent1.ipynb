```python
import requests
from dotenv import  load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.tools import  tool
from langchain.agents import  create_agent
```

```python
load_dotenv()
```

```python
llm=ChatGoogleGenerativeAI(
    model="gemini-3.5-flash",
    temperature=0.8
)
```

```python
api="https://681444af225ff1af16285b31.mockapi.io/furnitureapi"
apis="https://681444af225ff1af16285b31.mockapi.io/electronicapi"
```

```python
@tool
def mydataelectronics():
    """give me all electronics product from this api"""

    response=requests.get(api)
    return response.json()
```

```python
@tool
def mydatafurniture():
    """give me all furniture product from this api"""

    response=requests.get(apis)
    return response.json()
```

```python
result=mydataelectronics.invoke({})
result
```

```python
result1=mydatafurniture.invoke({})
result1
```

```python
agent=create_agent(
    model=llm,
    tools=[mydataelectronics,mydatafurniture],
    system_prompt="""
    you are a helpful assistant.

    use the electronis and futrniture tool whenever the user asks about
     available electronics and furniture products

     do not make us product information.
     use the tool data to answer.
    """
)
```

```python
question=input("ask me anything...")

result=agent.invoke({
    "messages":[("user",question)]
})

print(result["messages"][-1].content)
```
