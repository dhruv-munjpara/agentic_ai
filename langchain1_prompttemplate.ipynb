```python
from dotenv import load_dotenv
import os
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
# from langchain_core.messages import HumanMessage
from langchain_core.prompts import PromptTemplate
```

```python
load_dotenv()
```

```python
llm=ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0.10
)
```

```python
tech=PromptTemplate(
    input_variable=["tech"],
    template="""
    you are an experienced technical interviewer.
    
    Generate 5  inyerview questions for {tech}.
    
    Requirements:
    1.Generate 3 basic questions.
    2.Generate 2 adcvanced questions.
    3.keep the questions clear and practical.
     """
)
```

```python
technology=input("enter your question")
```

```python
prompt=tech.format(tech=technology)
```

```python
response=llm.invoke(prompt)
```

```python
print(response.content)
```
