```python
import requests
from dotenv import load_dotenv
from langchain_google_genai import  ChatGoogleGenerativeAI
from langchain_core.tools import  tool
from langchain.agents import  create_agent
```

```python
load_dotenv
```

```python
llm=ChatGoogleGenerativeAI(
    model="gemini-3.5-flash",
    temperature=0.8
)
```

```python
api="https://6a7dfc55f8b2ed99ca4f0a4a.mockapi.io/job_search"
```

```python
@tool
def job_searchdata():
    """give me all job data from thid api"""

    response=requests.get(api)
    return response.json()
```

```python
result=job_searchdata.invoke({})
result
```

```python
agent=create_agent(
    model=llm,
    tools=[job_searchdata],
    system_prompt="""
    you are a helpful assistant.

    use the job_search tool whenever the user asks about
    available job.

    do not make us job search information,
    use the tool data to answer
    """
)
```

```python
question=input("ask me for your job....")

result=agent.invoke({
    "messages":[("user",question)]
})

print(result["messages"][-1].content)
```
